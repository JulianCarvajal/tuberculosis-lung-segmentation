## 1 · Verificación del entorno

````python
from pathlib import Path

# Busco la raíz del proyecto subiendo desde el directorio actual.
# Si el notebook está en notebooks/, sube un nivel; si está en la raíz, se queda.
ROOT = next(
    (p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "tbseg").exists()),
    None,
)
if ROOT is None:
    raise RuntimeError(
        "No encontré la raíz del proyecto. "
        "Asegúrate de ejecutar este notebook desde dentro del repositorio clonado."
    )
print("Raíz del proyecto:", ROOT)

# Destinos de cada dataset
TBX11K_DIR = ROOT / "data" / "raw" / "TBX11K"
MS_DIR     = ROOT / "data" / "external" / "montgomery_shenzhen"

print("TBX11K irá en:              ", TBX11K_DIR)
print("Montgomery+Shenzhen irá en: ", MS_DIR)
````

## 2 · TBX11K

**Fuente:** [https://github.com/yun-liu/Tuberculosis](https://github.com/yun-liu/Tuberculosis)

El repositorio enlaza a Google Drive con los archivos del dataset.
Descarga el ZIP completo y extráelo de modo que la estructura quede así:

```
data/raw/TBX11K/
├── imgs/
│   ├── tb/
│   ├── health/
│   ├── sick/
│   ├── test/
│   └── extra/
├── lists/
│   ├── TBX11K_train.txt
│   ├── TBX11K_val.txt
│   └── TBX11K_trainval.txt
└── annotations/
    ├── xml/
    └── json/
```

Una vez extraído, ejecuta la celda siguiente para verificar que la estructura es correcta.

````python
# Verificación de TBX11K
expected_tbx = {
    "imgs/tb":     "imágenes TB",
    "imgs/health": "imágenes sanas",
    "imgs/sick":   "imágenes enfermas no-TB",
    "imgs/test":   "imágenes de test (sin etiqueta)",
    "imgs/extra":  "imágenes de fuentes externas",
    "lists":       "listas de partición oficial",
    "annotations/xml":  "anotaciones XML por imagen TB",
    "annotations/json": "anotaciones en formato COCO",
}

ok = True
for rel, descripcion in expected_tbx.items():
    path = TBX11K_DIR / rel
    if path.exists():
        n = len(list(path.iterdir()))
        print(f"  ✓  {rel:<30} {descripcion}  ({n} archivos)")
    else:
        print(f"  ✗  {rel:<30} NO ENCONTRADO")
        ok = False

print()
if ok:
    print("TBX11K está correctamente ubicado. Puedes continuar con el notebook 01b.")
else:
    print("Faltan carpetas. Revisa la extracción del ZIP y repite la verificación.")
````

## 3 · Montgomery + Shenzhen

**Fuente:** [https://www.kaggle.com/datasets/nikhilpandey360/chest-xray-masks-and-labels](https://www.kaggle.com/datasets/nikhilpandey360/chest-xray-masks-and-labels)

Descarga el ZIP desde Kaggle (requiere cuenta gratuita) y extráelo de modo que
la estructura quede así:

```
data/external/montgomery_shenzhen/
├── CXR_png/           radiografías en PNG

└── ClinicalReadings/  lecturas clínicas en TXT
```

Una vez extraído, ejecuta la celda siguiente para verificar que la estructura es correcta.

````python
# Verificación de Montgomery + Shenzhen
expected_ms = {
    "CXR_png":         "radiografías",
    "masks":           "máscaras pulmonares",
    "ClinicalReadings": "lecturas clínicas",
}

ok = True
for rel, descripcion in expected_ms.items():
    path = MS_DIR / rel
    if path.exists():
        n = len(list(path.iterdir()))
        print(f"  ✓  {rel:<25} {descripcion}  ({n} archivos)")
    else:
        print(f"  ✗  {rel:<25} NO ENCONTRADO")
        ok = False

print()
if ok:
    print("Montgomery + Shenzhen está correctamente ubicado. Puedes continuar con el notebook 01c.")
else:
    print("Faltan carpetas. Revisa la extracción del ZIP y repite la verificación.")
````
